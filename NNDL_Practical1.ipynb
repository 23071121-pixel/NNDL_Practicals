```python
import numpy as np

matrix1=np.array([[2,5,3],
                 [4,3,6]])
matrix2=np.array([[2,7],
                 [6,3]])

print("Matrix1=",matrix1)
print("Matrix2=",matrix2)

result=np.matmul(matrix2,matrix1)
print("Multiplication=",result)
```
